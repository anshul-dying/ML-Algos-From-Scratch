```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, r2_score
```

```python
df_classification = pd.read_csv('heart_classification.csv')
```

```python
df_classification.head()
```

```python
df_regression = pd.read_csv("Salary.csv")
```

```python
df_regression.head()
```

```python
from KNN import KNNClassifier, KNNRegression
```

```python
X_class = df_classification.iloc[:,:-1].values
y_class = df_classification.iloc[:,-1].values
X_reg = df_regression.iloc[:,:-1].values
y_reg = df_regression.iloc[:,-1].values
```

```python
k_class = KNNClassifier()
k_reg = KNNRegression()

k_class.fit(X_class, y_class)
k_reg.fit(X_reg, y_reg)

y_class_pred = k_class.predict(X_class)
y_reg_pred = k_reg.predict(X_reg)
```

```python
print(f"Classifier Accuracy: {accuracy_score(y_class, y_class_pred)*100:.2f} %")
print(f"Regressor Accuracy: {r2_score(y_reg, y_reg_pred)*100:.2f} %")
```
